```python
def stampa_griglia(tabellone):
    """Stampa la griglia di gioco aggiornata."""
    print("\n")
    print(f" {tabellone[0]} | {tabellone[1]} | {tabellone[2]} ")
    print("---|---|---")
    print(f" {tabellone[3]} | {tabellone[4]} | {tabellone[5]} ")
    print("---|---|---")
    print(f" {tabellone[6]} | {tabellone[7]} | {tabellone[8]} ")
    print("\n")


def controlla_vittoria(tabellone, giocatore):
    """Controlla se il giocatore attuale ha vinto."""
    combinazioni_vincenti = [
        # Righe
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        # Colonne
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        # Diagonali
        [0, 4, 8],
        [2, 4, 6],
    ]

    for comb in combinazioni_vincenti:
        if (
            tabellone[comb[0]]
            == tabellone[comb[1]]
            == tabellone[comb[2]]
            == giocatore
        ):
            return True
    return False


def gioco_tris():
    """Funzione principale del gioco."""
    # Inizializziamo il tabellone con numeri da 1 a 9 per indicare le posizioni
    tabellone = ["1", "2", "3", "4", "5", "6", "7", "8", "9"]
    giocatore_attuale = "X"
    mosse = 0

    print("--- BENVENUTO AL GIOCO DEL TRIS ---")
    stampa_griglia(tabellone)

    while mosse < 9:
        scelta = input(
            f"Giocatore {giocatore_attuale}, scegli una posizione (1-9): "
        )

        # Verifica validità dell'input
        if not scelta.isdigit() or int(scelta) < 1 or int(scelta) > 9:
            print(
                "❌ Inserimento non valido! Scegli un numero compreso tra 1 e 9."
            )
            continue

        indice = int(scelta) - 1

import random


def stampa_griglia(tabellone):
    """Stampa la griglia di gioco aggiornata."""
    print("\n")
    print(f" {tabellone[0]} | {tabellone[1]} | {tabellone[2]} ")
    print("---|---|---")
    print(f" {tabellone[3]} | {tabellone[4]} | {tabellone[5]} ")
    print("---|---|---")
    print(f" {tabellone[6]} | {tabellone[7]} | {tabellone[8]} ")
    print("\n")


def controlla_vittoria(tabellone, giocatore):
    """Controlla se il giocatore specificato ha vinto."""
    combinazioni_vincenti = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],  # Righe
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],  # Colonne
        [0, 4, 8],
        [2, 4, 6],  # Diagonali
    ]
    for comb in combinazioni_vincenti:
        if (
            tabellone[comb[0]]
            == tabellone[comb[1]]
            == tabellone[comb[2]]
            == giocatore
        ):
            return True
    return False


def posizioni_libere(tabellone):
    """Restituisce la lista degli indici ancora disponibili."""
    return [i for i, val in enumerate(tabellone) if val not in ["X", "O"]]


def mossa_cpu(tabellone, simbolo_cpu, simbolo_giocatore):
    """Strategia della CPU:
    1. Prova a vincere se ha una mossa decisiva.
    2. Blocca il giocatore se sta per vincere.
    3. Prende il centro (pos 5 / indice 4) se libero.
    4. Sceglie una posizione casuale tra quelle libere.
    """
    libere = posizioni_libere(tabellone)

    # 1. Controlla se la CPU può vincere in questa mossa
    for pos in libere:
        copia = tabellone.copy()
        copia[pos] = simbolo_cpu
        if controlla_vittoria(copia, simbolo_cpu):
            return pos

    # 2. Controlla se deve bloccare il giocatore
    for pos in libere:
        copia = tabellone.copy()
        copia[pos] = simbolo_giocatore
        if controlla_vittoria(copia, simbolo_giocatore):
            return pos

    # 3. Prendi il centro se disponibile
    if 4 in libere:
        return 4

    # 4. Altrimenti mossa casuale
    return random.choice(libere)


def gioco_tris_vs_cpu():
    tabellone = ["1", "2", "3", "4", "5", "6", "7", "8", "9"]
    giocatore_umano = "X"
    cpu = "O"
    turno = "X"  # Inizia l'umano

    print("=== TRIS: GIOCATORE (X) vs CPU (O) ===")
    stampa_griglia(tabellone)

    while len(posizioni_libere(tabellone)) > 0:
        if turno == giocatore_umano:
            # Turno del Giocatore
            scelta = input("Tuo turno! Scegli una posizione (1-9): ")

            if not scelta.isdigit() or int(scelta) < 1 or int(scelta) > 9:
                print("❌ Inserisci un numero valido da 1 a 9.")
                continue

            indice = int(scelta) - 1

            if tabellone[indice] in ["X", "O"]:
                print("⚠️ Casella già occupata! Scegli un'altra casella.")
                continue

            tabellone[indice] = giocatore_umano
            stampa_griglia(tabellone)

            if controlla_vittoria(tabellone, giocatore_umano):
                print("🎉 COMPLIMENTI! Hai battuto la CPU!")
                return

            turno = cpu

        else:
            # Turno della CPU
            print("🤖 La CPU sta pensando...")
            indice = mossa_cpu(tabellone, cpu, giocatore_umano)
            tabellone[indice] = cpu
            print(f"🤖 La CPU ha scelto la posizione {indice + 1}")
            stampa_griglia(tabellone)

            if controlla_vittoria(tabellone, cpu):
                print("💻 La CPU ha vinto! Ritenta, sarai più fortunato.")
                return

            turno = giocatore_umano

    print("🤝 Pareggio! Nessuno ha vinto.")


if __name__ == "__main__":
    gioco_tris_vs_cpu()
```
